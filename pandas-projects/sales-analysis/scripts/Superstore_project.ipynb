```python
import pandas as pd
df = pd.read_csv('superstore_data.csv', encoding='latin1')
```

```python
df.head(10)

##Total Sales
df.loc[:,'Sales'].sum(axis=0)

df['Sales'].sum() #Easier way to get the total sales
```

```python
##Top 10 Products with the highest sales
df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10) #A series with one column

df.groupby('Product Name')[['Sales', 'Quantity']].sum().sort_values('Sales', ascending=False).head(10) #A dataframe with multiple columns

df.groupby('Product Name').agg({'Sales': 'sum', 'Quantity': 'sum'}).sort_values('Sales', ascending=False).head(10)
```

```python
##Which category makes the most money

df.groupby('Category')['Sales'].sum().sort_values(ascending=False).head(10)
```

```python
##Which region performs the best
df.groupby('State')['Sales'].sum().sort_values(ascending=False).head(10)
```

```python
## Business Summary 
# Total Revenue 
# Best Category 
# Best Region 
# Top Product 


def total_revenue(df):
    return df['Sales'].sum()
def best_category(df):
    return df.groupby('Category')['Sales'].sum().idxmax()

def best_region(df):
    return df.groupby('Region')['Sales'].sum().idxmax()

def top_product(df):
    return df.groupby('Product Name')['Sales'].sum().idxmax()

def business_summary(df):
    total_revenue = df['Sales'].sum()
    best_category = df.groupby('Category')['Sales'].sum().idxmax()
    best_region = df.groupby('Region')['Sales'].sum().idxmax()
    top_product = df.groupby('Product Name')['Sales'].sum().idxmax()
    
    summary = {
        'Total Revenue': total_revenue,
        'Best Category': best_category,
        'Best Region': best_region,
        'Top Product': top_product
    }
    
    return summary

print("\nWhat would you like to get?")
print("1. Total Revenue")
print("2. Best Category")
print("3. Best Region")
print("4. Top Product")
print("5. Business Summary")

choice = input("Enter your choice (1-5): ")

if choice == '1':
    print("Total Revenue: ", total_revenue(df))
elif choice == '2':
    print("Best Category:", best_category(df))
elif choice == '3':
    print("Best Region:", best_region(df))
elif choice == '4':
    print("Top Product:", top_product(df))
elif choice == '5':
    summary = business_summary(df)
    for key, value in summary.items():
        print(f"{key}: {value}")      #The f word is used to format the string and insert the values of the variables into the string.    
else:
    print("Invalid input. Please enter a number between 1 and 5.")



```
